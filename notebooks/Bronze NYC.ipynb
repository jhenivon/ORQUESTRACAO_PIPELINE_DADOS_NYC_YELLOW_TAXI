## Amostra de Dados na Bronze

```sql
%python
from pyspark.sql import functions as F
df_bronze = spark.read.parquet(
    "abfss://data@adlsdatalakehousedevbr.dfs.core.windows.net/bronze/nyc_taxi/NYC_TAXI"
)

display(df_bronze.limit(10) )
```

```sql
%python
df_bronze.createOrReplaceTempView("tbl_nyc_taxi")
```

## Amostra de dados depois da tabela criada

```sql
SELECT * FROM tbl_nyc_taxi LIMIT 5
```

## Salvando Camada Bronze na Silver

```sql
%python
silver_path = "abfss://data@adlsdatalakehousedevbr.dfs.core.windows.net/silver/nyc_taxi"

df_bronze.write \
    .mode("overwrite") \
    .parquet(silver_path)
```
