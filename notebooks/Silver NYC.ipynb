```python
from pyspark.sql import functions as F
```

## Amostra de Dados na Silver

```python
df_silver = spark.read.parquet("abfss://data@adlsdatalakehousedevbr.dfs.core.windows.net/silver/nyc_taxi")
display(df_silver.limit(10))
```

## Transformacoes de datas e outra colunas para valores nulos

```python
df_silver = (
    df_silver
    .withColumn("tpep_pickup_datetime", F.to_timestamp("tpep_pickup_datetime") )
    .withColumn("tpep_dropoff_datetime", F.to_timestamp("tpep_dropoff_datetime") )
    .fillna({
        "passenger_count": 1,
        "RatecodeID": 1,
        "store_and_fwd_flag": "N",
        "congestion_surcharge": 2.5,
        "Airport_fee": 0
    })
)
display(df_silver.limit(10))
```

## Salvando df_silver na camada Gold como df_gold

```python
gold_path = "abfss://data@adlsdatalakehousedevbr.dfs.core.windows.net/gold_analytics/nyc_taxi"

df_silver.write \
    .mode("overwrite") \
    .parquet(gold_path)
```
